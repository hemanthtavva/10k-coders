```python
#anagaram check
s1 = "pot"
s2 = "top"
is_anagram = True
if len(s1) == len(s2):
    for char in s1 :
        if s1.count(char) != s2.count(char):
            is_anagram= False
            break
else :
    is_anagram = False

print(is_anagram)
```

```python
#accesing the json data
details = [
    {"name":"Hemanth", "marks":[50,70,100]},
    {"name":"mahi", "marks":[90,70,100]},
    {"name":"Hema", "marks":[50,80,100]}

]
stu_name = "" 
max_marks = 0
for stu in details :
    total_marks = sum(stu["marks"])
    if total_marks > max_marks:
        max_marks = total_marks
        stu_name = stu["name"]
print(stu_name)
```

```python
#finding the toppers in each subject
details = [
    {"name":"Hemanth", "marks":{"tel":50,"hin":70,"eng":100}},
    {"name":"mahi", "marks":{"tel":90,"hin":70,"eng":97}},
    {"name":"Hema", "marks":{"tel":50,"hin":80,"eng":90}}

]
report={}
tel = 0
hin = 0
eng = 0
for stu in details:
    for key in stu["marks"]:
        if key == "tel":
            if stu["marks"][key]>tel:
                tel = stu["marks"][key]
                report[key] = stu["name"] 

        if key == "hin":
            if stu["marks"][key]>hin:
                hin = stu["marks"][key]
                report[key] = stu["name"] 

        if key == "eng":
            if stu["marks"][key]>eng:
                eng = stu["marks"][key]
                report[key] = stu["name"] 
print(report)
```

```python
#we have to re arrange the array such that all 0's are sent to end of the list and the order of the other elements should be maintained
a = [1, 0, 3, 0, 2, 4, 0, 5]
s = 0
for i in range(len(a)):
    if a[i] != 0:
        a[s], a[i] = a[i], a[s]
        s += 1

print(a)
```

```python
nums = [2,7,3,6,4,5]
n=3
cur = sum(nums[:n])
maxi=cur
for i in range(0,len(nums)-n):
    cur = cur-nums[i]+nums[i+n]
    if cur>maxi:
        maxi = cur
print(maxi)
```

```python
nums = [2,4,1,7,3,6]
target = 9
unq = set()
for i in range(len(nums)):
    if target-nums[i] in unq:
        print(nums[i],target-nums[i])
    else:
        unq.add(nums[i])
```

```python
nums = [2,4,1,7,3,6]
target = 9
s=0
l=len(nums)-1
while s<l:
    if nums[s]+nums[l] == target:
        print(nums[s],nums[l])
        break
    elif nums[s]+nums[l] < target:
        s+=1
    else:
        l-=1
```
