```python
nums = np.random.randint(1,100,25)
print(nums)
print(nums.mean())
```

```python
#min
nums = np.random.randint(30,77,5)
print(nums)
print(nums.min())
```

```python
#max
nums = np.random.randint(55,99,13)
print(nums)
print(nums.max())
```

```python
#standard deviation
marks = np.random.randint(70,100,15)
print(marks)
print(marks.std())
print(marks.mean())
```

```python
#variance
nums = np.random.randint(44,99,11)
print(nums)
print(nums.var())
```

```python
#argmax usedto find the position of maximum value in the array
nums = np.random.randint(22,44,9)
print(nums)
print(nums.argmax())
```

```python
#argmin used to find the position of minimum value in the array
nums = np.random.randint(22,44,9)
print(nums)
print(nums.argmin())
```

```python
#sqrt used to find the square root of the elements in the array
nums = np.random.randint(1,50,10)
print(nums)
print(np.sqrt(nums))
```

```python
#pow is used to find the poer of the elements in the array by giving the how much power we wanted from the elements 
nums = np.random.randint(1,50,10)
print(nums)
print(np.pow(nums,3))
```

```python
#round function is used to round the float values in the array
nums = np.random.rand(1,4,10)
print(nums)
print(np.round(nums))
```

```python
#abs is used to turn the negative values in the array to positive values
nums = np.array([-1,2,3,-4])
print(nums)
print(np.abs(nums))
```

```python
#sort is used to sort the values in the array into ascending order
nums = np.random.randint(1,50,10)
print(nums)
print(np.sort(nums))
```

```python
#sort is used to sort the values in the array into decending order
nums = np.random.randint(1,50,10)
print(nums)
sortt = np.sort(nums) [::-1]
print(sortt)
```

```python
#where is used to find the elements postion based on the condition we have given
nums = np.array([1,2,5,77,3,44,6,22])
print(np.where(nums > 30))
```

```python
#unique is used to remove the repeated or dupilicate values 
nums = np.array([1,2,3,44,2,5,77,3,44,6,22])
print(np.unique(nums))
```
