```python
df = pd.DataFrame({
    "org_name": ["google", "oracle", "infosys", "jp morgan", "deloitte"],
    "dept": ["IT", "HR", "SALES", "MARKETING", "MANAGING"],
    "type": ["product", "product", "service", "service", "service"],
    "salary": [100000, 80000, 50000, 90000, 35000]
}, index=["c1", "c2", "c3", "c4", "c5"])

```

```python
print(df)
```

```python
#boolean filtering
sal = df["salary"] > 50000
print(sal)
print(df[sal])
```

```python
#Boolean - AND
con = (df["dept"] == 'IT') & (df["salary"] > 50000)
print(df[con])
```

```python
#Boolean - OR
con = (df["dept"] == 'IT') | (df["salary"] > 50000)
print(df[con])
```

```python
#Boolean - NOT
con = df[~(df["dept"] == 'IT') ]
print(con)
```

```python
#Boolean - Greater Than
con = df["salary"] > 50000
print(df[con])
```

```python
#Boolean - less Than
con = df["salary"] < 50000
print(df[con])
```
