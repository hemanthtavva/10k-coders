### Passenger Count Missing-Value Treatment

There were 20 missing values in Passenger_Count. The affected
records were distributed across multiple routes, with no obvious
single-route pattern.

Instead of removing these trips, missing passenger counts were
imputed using the median Passenger_Count of the corresponding
Route_ID. Route-level median imputation was selected because
passenger demand can vary by route and the median is less sensitive
to extreme passenger counts than the mean.

After imputation, Passenger_Count contains no missing values and
the original 20,000 trip records were retained.

### Zero Distance Treatment

There were 10 trips with Distance_KM equal to zero. These records
had different origin and destination stations, so a zero distance
was treated as invalid.

The invalid distances were replaced using the median Distance_KM
for the corresponding Route_ID, calculated only from valid
positive-distance trips.

Route-level median imputation was selected because trip distances
can vary by route, while the median reduces the influence of
unusually short or long trips.

No trip records were removed because of the invalid distance values.
