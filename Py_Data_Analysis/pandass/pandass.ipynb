```python
import pandas as pd
print(pd.__version__)
```

```python
# series is a basic one dimensional labeled collection of data in pandas
# the data looks like it has index and the related value to it
marks = pd.Series([85,90,79,99,100])
print(marks)
```

```python
#in series we can give the custom indexes also
marks = pd.Series([85,90,88,99], index=["python","SQL","Excel","PowerBI"])
print(marks)
```

```python
#we can access th data using the custom indexes also 
print(marks["python"])
```

```python
# data frame is basically a table, it has collection of series with same index
emp = {
    "Name" :["Hemanth","karthik", "Ganesh"],
    "Age" : [21,25,22],
    "salary" : [40000,50000,60000]
}
df = pd.DataFrame(emp)
print(df)
```

```python
df = pd.DataFrame( 
    emp,
    index = [101,102,103]
)
print(df)
```
