```python
import numpy as np
```

```python
#numpy zeroes()
# it creates a array filled with xeroes, by default it is stored in float value,
#but we can modify it by declaring the data type,
# it is useful in some cases example:- 
```

```python
arr = np.zeros(5)
print(arr)
```

```python
arrs = np.zeros(5, dtype = int)
print(arrs)
print(arr.shape)
```

```python
arrs = np.zeros((5,5),dtype='int64')
print(arrs)
print(arr.shape)
```

```python
#numpy ones()
# it creates a array filled with ones, by default it is stored in float value,
#but we can modify it by declaring the data type,
```

```python
arrs = np.ones((5,5), dtype=int)
print(arrs)
```

```python
#numpy identity()
# it creates a array filled with ones with the diagonal column, by default it is stored in float value,
#but we can modify it by declaring the data type,
```

```python
arrs = np.identity((5),dtype= int)
print(arrs)
```

```python
arrs = np.eye((5), dtype=int)
print(arrs)
```

```python
#the diff bertween the identity and eye is eye can do regular matrixes also 
```

```python
arrs = np.eye(3,5, dtype=int)
print(arrs)
```

```python
#empty is used to declare thezeroes like the zeroes function but it is empty 
```

```python
arr = np.empty((4,4))
print(arr)
```

```python
#full
arrs = np.full(3,5)
print(arrs)
```

```python
#arange
# nums = ( start, stop, step)
```

```python
nums = np.arange(1,10)
print(nums)
```

```python
nums = np.arange(1,10,2)
print(nums)
```

```python
nums = np.arange(10,0,-1)
print(nums)
```

```python
#linspace
# nums = (start, stop , number of values)
```

```python
arr = np.linspace(1,10,5, dtype=int)
print(arr)
```

```python
# the difference bwtween the arange and the linsapce is that 
#arange controls the step size and the #linspace will comtrol how many values have to return, 
# like homany parts we want to make the range of nums
```

```python
#random
#nums = (start, stop, how many random numbers to generate)
#uses rand for decimal values
# uses randint for onteger values
```

```python
nums = np.random.randint(1,100,5)
print(nums)
```
