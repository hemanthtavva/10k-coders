```python
from datetime import datetime, date, time, timedelta
```

```python
#to get current date time 
print(datetime.now())
```

```python
#to get current date
print(date.today())
```

```python
my_bday = date(2004,12,29)
print(my_bday)
```

```python
age = date.today() - my_bday
print(age)
```

```python
b_time = time(12, 30, 45)
print(b_time)
```

```python
#timedelta is used to calculate the difference between two dates or time.
# it can also be use to add or subtract days, weeks, months, years from a date or time.
sub_date = date.today()
expires_on = sub_date + timedelta(days=30)
print(expires_on)
```

```python
#usecase of timedelta
login_time = datetime(2026,7,4,11,15,0)
emp_login_time = datetime.now() 
grace_period = timedelta(minutes=15)
if emp_login_time > login_time + grace_period:
    print("Employee is late")
else:
    print("Employee is on time")
```
