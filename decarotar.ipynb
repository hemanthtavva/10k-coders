```python
#decorators are functions that adds functionality to the existing functions without modifying their structure.
#  They are often used for logging, access control, memoization, and more.
```

```python
def argss(*args, **kwargs):
    return args, kwargs

print(argss( name="hemanrth"))
print(argss(1,2))
```

```python
def check_int(func):
    def wrapper(*args, **kwargs):
        for i in args:
            if not isinstance(i, int):
                return "Invalid input"
        return func(*args, **kwargs)
    return wrapper


@check_int
def add(a, b):
    return a + b


print(add(1, "2"))  
print(add(1, 2))    
```

```python
@check_int
def mul(a, b):
    return a * b
print(mul(1, '2'))
```

```python
def valid_report(func):
    def wrapper(*args, **kwargs):
        if len(args) != 6:
            return " enter vaild marks"
        for i in args:
            if not isinstance(i, int):
                return "Invalid input"
        if kwargs.get("name") is None:
            return "Name is required"
        return func(*args, **kwargs)
    return wrapper

@valid_report
def report_card(*args, **kwargs):
    return f"retport card for {kwargs.get('name')}:{sum(args)}"
print(report_card(90, 80, 70, 60, 50, 40, name="Hemanth"))
```

```python
def exe_time(func):
    def wrapper(*args, **kwargs):
        start_time = datetime.now()
        
        end_time = datetime.now()
        execution_time = end_time - start_time
        return execution_time
        
    return wrapper


@exe_time
def sample_function():
    total = 0
    for i in range(1000000):
        total += i
    return total
sample_function()

```
